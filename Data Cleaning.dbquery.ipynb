```sql
Create or replace view fraud_transactions_for_detection AS

with exact_duplicates_removed AS (
    Select distinct * 
    from fraud_detection_dataset
),

version_counts as (
    select
        * ,
        count(*) over (
            partition by Transaction_ID
            ) as versions_per_transaction
        from exact_duplicates_removed
)
        
select * from version_counts where versions_per_transaction = 1;
```
