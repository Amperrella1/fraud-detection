```sql
SELECT *
FROM fraud_detection_dataset
LIMIT 10;
DESCRIBE TABLE fraud_detection_dataset;

SELECT 
    count(*) as total_rows,
    count(DISTINCT Transaction_ID) AS unique_transactions,
    count(DISTINCT User_ID) as unique_users
from fraud_detection_dataset;


select
    Transaction_ID,
    User_ID,
    Transaction_Amount,
    Transaction_Type,
    Fraudulent
FROM fraud_detection_dataset
WHERE Fraudulent = 1
ORDER BY Transaction_Amount DESC
LIMIT 10;

SELECT
    Fraudulent,
    Count (*) as transaction_count
From fraud_detection_dataset
Group by Fraudulent
Order by Fraudulent;


    

```
